```python
df["discounted_price"] = (df["discounted_price"].str.replace("₹","").str.replace(",","").astype(float))
```

```python
df["actual_price"] = (df["actual_price"].str.replace("₹","").str.replace(",","").astype(float))
```

```python
df["discount_percentage"] = (df["discount_percentage"].str.replace("%","").astype(float))
```

```python
df["rating"] = pd.to_numeric(df["rating"],errors="coerce")
```

## Section 5: Feature Engineering

```python
df["savings"] = (df["actual_price"]- df["discounted_price"])
```

```python
df["price_ratio"] = (df["discounted_price"]/ df["actual_price"])
```

```python
# Create classification target:
df["high_rating"] = (df["rating"] >= 4).astype(int)
```

## Section 6: Machine Learning

```python
x = df[ ["actual_price","discounted_price","discount_percentage","savings","price_ratio"]]
y = df["high_rating"]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## Section 7: Model Comparison

```python
# Logistic regression
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()

lr.fit(X_train,y_train)

lr_score = lr.score(X_test,y_test)
```

```python
# KNN

from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier()

knn.fit(X_train,y_train)

knn_score = knn.score(X_test,y_test)
```

```python
# Decision Tree
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier()

dt.fit(X_train,y_train)

dt_score = dt.score(X_test,y_test)
```

```python
# Random Forest
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()

rf.fit(X_train,y_train)

rf_score = rf.score(X_test,y_test)
```

```python
results = pd.DataFrame({
    "Model":[
        "Logistic Regression",
        "KNN",
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy":[
        lr_score,
        knn_score,
        dt_score,
        rf_score
    ]
})

results.sort_values(
    "Accuracy",
    ascending=False
)
```

## Key Findings

1. Most products have ratings above 4.
2. Discounts alone do not guarantee higher ratings.
3. Actual price has weak correlation with ratings.
4. Savings and discount percentage contribute to prediction.
5. Random Forest achieved the highest accuracy.
